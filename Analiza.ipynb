```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = r"moje_wyniki.csv"
df = pd.read_csv(file_path, encoding="utf-8")

df_miasta = df.head(120)
df_miasta.to_csv("wyniki_miasta.csv", index=False, encoding="utf-8")
```

```python
sns.set(style="whitegrid", palette="muted", font_scale=1.1)

plt.figure(figsize=(14, 5))
plt.subplot(1, 3, 1)
sns.boxplot(data=df_miasta, x="LiczbaMiast", y="Czas", palette="Set1")
plt.title("Średni czas działania vs liczba miast")
plt.ylabel("Czas (s)")
plt.xlabel("Liczba miast")
plt.xticks(rotation=45)


plt.subplot(1, 3, 2)
sns.boxplot(data=df_miasta, x="LiczbaMiast", y="Fitness", palette="Set2")
plt.title("Średni Fitness vs liczba miast")
plt.ylabel("Fitness")
plt.xlabel("Liczba miast")
plt.xticks(rotation=45)


plt.subplot(1, 3, 3)
sns.boxplot(data=df_miasta, x="LiczbaMiast", y="OstatniaGeneracja", palette="Set3")
plt.title("Średnia wartość ost. generacji vs liczba miast")
plt.ylabel("Ostatnia Generacja")
plt.xlabel("Liczba miast")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df_miasta, x="OstatniaGeneracja", y="Fitness", hue="LiczbaMiast", palette="coolwarm", alpha=0.7)
plt.title("Fitness vs Ostatnia Generacja")
plt.xlabel("Ostatnia Generacja")
plt.ylabel("Fitness")
plt.legend(title="Liczba miast", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
grouped = df_miasta.groupby("LiczbaMiast").agg({
    "Fitness": "std",
    "OstatniaGeneracja": "std",
}).reset_index()

grouped.columns = ["LiczbaMiast", "StdFitness", "StdGeneracja"]

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
sns.lineplot(data=grouped, x="LiczbaMiast", y="StdFitness", marker="o")
plt.title("Wariancja Fitness vs Liczba Miast")
plt.xlabel("Liczba miast")
plt.ylabel("Odchylenie standardowe Fitness")

plt.subplot(1, 2, 2)
sns.lineplot(data=grouped, x="LiczbaMiast", y="StdGeneracja", marker="o", color="green")
plt.title("Wariancja Generacji vs Liczba Miast")
plt.xlabel("Liczba miast")
plt.ylabel("Odchylenie standardowe Ostatniej Generacji")

plt.tight_layout()
plt.show()
```

```python
df_selections = pd.concat([
    df[df["PopulationID"] == "00dac625-350d-4502-8a76-2d12453a64b6"],
    df[df["PopulationID"] == "28ed13df-f0a6-4242-abb8-26965da841bf"],
    df[df["PopulationID"] == "52d51980-7ecb-4331-8a6e-9934d868fd5f"]
]).dropna()
df_selections.to_csv("wyniki_selekcje.csv", index=False, encoding="utf-8")
```

```python
plt.figure(figsize=(16, 6))
sns.boxplot(data=df_selections, x="Selekcja", y="Fitness", hue="LiczbaMiast")
plt.title("Rozkład Fitness dla różnych selekcji i liczby miast")
plt.xlabel("Selekcja")
plt.ylabel("Fitness")
plt.legend(title="Liczba miast")
plt.show()
```

```python
grouped = df_selections.groupby(["LiczbaMiast", "Selekcja"]).agg({
    "Fitness": "mean",
    "Czas": "mean",
    "OstatniaGeneracja": "mean"
}).reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=grouped, x="LiczbaMiast", y="Fitness", hue="Selekcja", marker="o")
plt.title("Średni Fitness w zależności od selekcji i liczby miast")
plt.xlabel("Liczba miast")
plt.ylabel("Średni Fitness")
plt.show()
```

```python
grouped_std = df_selections.groupby(["LiczbaMiast", "Selekcja"]).agg({
    "Fitness": "std"
}).reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(data=grouped_std, x="Selekcja", y="Fitness", hue="LiczbaMiast")
plt.title("Odchylenie standardowe Fitnessu dla różnych selekcji")
plt.xlabel("Selekcja")
plt.ylabel("Odchylenie standardowe Fitness")
plt.show()
```

```python
g = sns.FacetGrid(df_selections, col="LiczbaMiast", height=5, sharey=False)
g.map_dataframe(sns.boxplot, x="Selekcja", y="Fitness")
g.set_titles("{col_name} miast")
g.set_axis_labels("Selekcja", "Fitness")
g.tight_layout()
plt.show()
```

```python
grouped_time = df_selections.groupby(["Selekcja", "LiczbaMiast"]).agg({"Czas": "mean"}).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=grouped_time, x="Selekcja", y="Czas", hue="LiczbaMiast")
plt.title("Średni czas działania algorytmu dla różnych selekcji i liczby miast")
plt.ylabel("Średni czas działania [s]")
plt.xlabel("Metoda selekcji")
plt.legend(title="Liczba miast")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
grouped_gen = df_selections.groupby(["Selekcja", "LiczbaMiast"]).agg({"OstatniaGeneracja": "mean"}).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=grouped_gen, x="Selekcja", y="OstatniaGeneracja", hue="LiczbaMiast")
plt.title("Średnia liczba generacji do zakończenia dla różnych selekcji")
plt.ylabel("Średnia liczba generacji")
plt.xlabel("Metoda selekcji")
plt.legend(title="Liczba miast")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
grouped_fitness = df_selections.groupby(["Selekcja", "LiczbaMiast"]).agg({"Fitness": "mean"}).reset_index()

pivot_fitness = grouped_fitness.pivot(index="LiczbaMiast", columns="Selekcja", values="Fitness")

plt.figure(figsize=(8, 6))
sns.heatmap(pivot_fitness, annot=True, fmt=".1f", cmap="viridis")
plt.title("Średni Fitness w zależności od selekcji i liczby miast")
plt.ylabel("Liczba miast")
plt.xlabel("Metoda selekcji")
plt.show()
```

```python
df.tail(240).to_csv("wyniki_cros_mut.csv", index=False, encoding="utf-8")

df_cm = pd.read_csv("wyniki_cros_mut.csv", encoding="utf-8")



df_cm['Krzyżowanie_Mutacja'] = df_cm['Krzyżowanie'] + ' + ' + df_cm['Mutacja']

plt.figure(figsize=(14,8))
sns.boxplot(data=df_cm, x='Krzyżowanie_Mutacja', y='Fitness', hue='LiczbaMiast')
plt.title('Porównanie Fitness w zależności od metody krzyżowania i mutacji')
plt.xticks(rotation=45)
plt.ylabel('Wartość funkcji celu (Fitness)')
plt.xlabel('Metoda Krzyżowania + Mutacja')
plt.legend(title='Liczba Miast')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,8))
sns.barplot(data=df_cm, x='Krzyżowanie_Mutacja', y='Czas', hue='LiczbaMiast', ci="sd")
plt.title('Średni czas wykonania dla różnych metod')
plt.xticks(rotation=45)
plt.ylabel('Czas (s)')
plt.xlabel('Metoda Krzyżowania + Mutacja')
plt.legend(title='Liczba Miast')
plt.tight_layout()
plt.show()
```
